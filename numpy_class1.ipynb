```python
import numpy as np

arr = np.array([1, 2, 3])       # int64 by default
print(arr.dtype)                 # int64

new_arr = arr.astype(float)     # convert to float
print(new_arr)                   # [1. 2. 3.]
print(new_arr.dtype)             # float64
```

```python
arr = np.array([1 ,2 ,3 ,4,6 ,5])

print(arr.max())
print(arr.min())
print(arr.sum())
print(arr.mean())
print(np.median(arr))
print(arr.std())
print(arr.var())
print(arr.argmin())
print(arr.argmax())

```
