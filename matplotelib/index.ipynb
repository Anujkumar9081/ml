<h2 style="color:#f97316;">4. Univariate Analysis — One Variable</h2>

<h3>Histogram</h3>

<pre><code>sns.histplot(data=df, x="Age")</code></pre>

<p><b>Used for:</b> Distribution of a numerical column.</p>

<p>Example question: <i>How are passenger ages distributed?</i></p>

<h3>Countplot</h3>

<pre><code>sns.countplot(data=df, x="Survived")</code></pre>

<p><b>Used for:</b> Count of categories.</p>

<p>Example: <code>0 = did not survive</code>, <code>1 = survived</code>.</p>

<hr>

<h2 style="color:#f97316;">5. Bivariate Analysis — Two Variables</h2>

<h3>Boxplot</h3>

<pre><code>sns.boxplot(data=df, x="Survived", y="Fare")</code></pre>

<p><b>Used for:</b> Compare a numerical variable between groups and identify outliers.</p>

<h3>Scatterplot</h3>

<pre><code>sns.scatterplot(data=df, x="Age", y="Fare")</code></pre>

<p><b>Used for:</b> See the relationship between two numerical variables.</p>

<hr>

<h2 style="color:#f97316;">6. <code>hue=</code> ⭐</h2>

<pre><code>sns.scatterplot(
    data=df,
    x="Age",
    y="Fare",
    hue="Survived"
)</code></pre>

<p><b>Meaning:</b> Add another variable to separate the observations into categories.</p>

<p>Here, <code>Survived</code> helps us visually compare survivors and non-survivors.</p>

<hr>

<h2 style="color:#7c3aed;">7. Multivariate Analysis — Many Variables</h2>

<h3>Pairplot</h3>

<pre><code>sns.pairplot(
    df[["Age", "Fare", "Pclass", "Survived"]],
    hue="Survived"
)</code></pre>

<p><b>Used for:</b> Quickly see relationships between multiple numerical features.</p>

<h3>Heatmap ⭐</h3>

<pre><code>sns.heatmap(
    df.corr(numeric_only=True),
    annot=True
)</code></pre>

<p><b>Used for:</b> Visualize correlations between numerical columns.</p>

<p><b>Correlation:</b></p>

<pre><code>-1  ← opposite relationship
 0  ← little/no linear relationship
+1  ← same relationship</code></pre>
